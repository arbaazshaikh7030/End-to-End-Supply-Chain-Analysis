```python
import pandas as pd

df = pd.read_csv(
    "Arbaaz BA Ruchi Mishra project.csv"
)

print("Dataset loaded successfully")

print("Number of rows:", df.shape[0])

print("Number of columns:", df.shape[1])

display(df.head())
```

```python
print("Dataset Information:")

df.info()

print("\nColumn Names:")

print(df.columns.tolist())

print("\nDataset Shape:")

print(df.shape)
```

```python
missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)

print(
    "\nTotal Missing Values:",
    df.isnull().sum().sum()
)
```

```python
print(
    "Duplicate Rows:",
    df.duplicated().sum()
)
```

```python
display(df.describe())
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)

print(df.columns.tolist())
```

```python
df = df.drop_duplicates()

print("Rows after cleaning:", len(df))
```

```python
numeric_columns = [
    "Price",
    "Availability",
    "Number of products sold",
    "Revenue generated",
    "Stock levels",
    "Lead times",
    "Order quantities",
    "Shipping times",
    "Shipping costs",
    "Lead time",
    "Production volumes",
    "Manufacturing lead time",
    "Manufacturing costs",
    "Defect rates",
    "Costs"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

print(df[numeric_columns].dtypes)
```

```python
print("Missing Values:")

print(
    df.isnull().sum().sort_values(
        ascending=False
    )
)

print("\nData Types:")

print(df.dtypes)
```

```python
df["shipping_time_category"] = pd.cut(
    df["Shipping times"],
    bins=[-float('inf'), 2, 5, float('inf')],
    labels=[
        "Fast",
        "Medium",
        "Slow"
    ]
)

display(
    df[
        [
            "SKU",
            "Shipping times",
            "shipping_time_category"
        ]
    ].head(10)
)
```

```python
import numpy as np

df["inventory_status"] = np.where(
    df["Stock levels"] < df["Order quantities"],
    "Stock Below Order Quantity",
    "Stock At or Above Order Quantity"
)

display(
    df[
        [
            "SKU",
            "Stock levels",
            "Order quantities",
            "inventory_status"
        ]
    ].head(10)
)
```

```python
print("Final Rows:", df.shape[0])

print("Final Columns:", df.shape[1])

display(df.head())
```

```python
import matplotlib.pyplot as plt

revenue_by_product = (
    df.groupby("Product type")["Revenue generated"]
    .sum()
    .sort_values(ascending=False)
)

revenue_by_product.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Revenue by Product Type")

plt.xlabel("Product Type")

plt.ylabel("Revenue Generated")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
shipping_cost_analysis = (
    df.groupby("transportation_modes")["shipping_costs"]
    .mean()
    .sort_values(ascending=False)
)

shipping_cost_analysis.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title(
    "Average Shipping Cost by Transportation Mode"
)

plt.xlabel("Transportation Mode")

plt.ylabel("Average Shipping Cost")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
stock_analysis = (
    df.groupby("product_type")["stock_levels"]
    .mean()
    .sort_values(ascending=False)
)

stock_analysis.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Average Stock Level by Product Type")

plt.xlabel("Product Type")

plt.ylabel("Average Stock Level")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```
