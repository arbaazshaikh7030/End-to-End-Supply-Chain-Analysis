```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files

uploaded = files.upload()

df = pd.read_csv("supply_chain_cleaned.csv")

print("Dataset shape:", df.shape)

df.head()
```

```python
print("Dataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Let's check the columns of df first to make sure they match perfectly
# Correcting 'Lead time' to 'Lead times' which we can see in df.head()

supplier_analysis = (
    df.groupby("Supplier name")
      .agg(
          Average_Lead_Time=("Lead times", "mean"),
          Average_Defect_Rate=("Defect rates", "mean"),
          Average_Manufacturing_Cost=("Manufacturing costs", "mean"),
          Record_Count=("SKU", "count")
      )
      .round(2)
      .sort_values("Average_Lead_Time")
)

supplier_analysis
```

```python
inventory_analysis = (
    df.groupby("Product type")
      .agg(
          Average_Stock=("Stock levels", "mean"),
          Total_Products_Sold=("Number of products sold", "sum"),
          Average_Availability=("Availability", "mean")
      )
      .round(2)
      .sort_values("Total_Products_Sold", ascending=False)
)

inventory_analysis
```

```python
inventory_analysis["Total_Products_Sold"].plot(kind="bar")

plt.title("Products Sold by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Products Sold")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
shipping_analysis = (
    df.groupby("Shipping carriers")["Shipping costs"]
      .mean()
      .sort_values(ascending=False)
)

shipping_analysis.plot(kind="bar")

plt.title("Average Shipping Cost by Carrier")
plt.xlabel("Shipping Carrier")
plt.ylabel("Average Shipping Cost")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
shipping_analysis = (
    df.groupby("Shipping carriers")["Shipping costs"]
      .mean()
      .sort_values(ascending=False)
)

shipping_analysis.plot(kind="bar")

plt.title("Average Shipping Cost by Carrier")
plt.xlabel("Shipping Carrier")
plt.ylabel("Average Shipping Cost")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
shipping_time_analysis = (
    df.groupby("Transportation modes")["Shipping times"]
      .mean()
      .sort_values(ascending=False)
)

shipping_time_analysis.plot(kind="bar")

plt.title("Average Shipping Time by Transportation Mode")
plt.xlabel("Transportation Mode")
plt.ylabel("Average Shipping Time")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
supplier_analysis.to_csv("supplier_analysis.csv")
inventory_analysis.to_csv("inventory_analysis.csv")
shipping_analysis.to_csv("shipping_cost_analysis.csv")
shipping_time_analysis.to_csv("shipping_time_analysis.csv")

print("Analysis files created successfully.")
```
